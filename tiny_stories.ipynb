```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

tokenizer = AutoTokenizer.from_pretrained("EleutherAI/gpt-neo-125M")

model_names = [
    "roneneldan/TinyStories-1M",
    "roneneldan/TinyStories-8M",
    "roneneldan/TinyStories-33M",
]

# Each case: a two-entity prompt ending right before the critical pronoun's
# referent-revealing continuation, plus the "correct" and "incorrect" next word.
# Design these so the two candidate words are both plausible completions,
# differing only in which entity they refer back to.
cases = [
    {
        "prompt": "Tom was a boy. Anna was a girl. Tom went to the park because",
        "correct": " he",
        "incorrect": " she",
    },
    {
        "prompt": "The dog was tired. The cat was hungry. The cat went to the bowl because",
        "correct": " it",   # ambiguous on its own -- fine as a sanity check, not a real test case
        "incorrect": " they",
    },
    {
        "prompt": "Sam was happy. Ben was sad. Ben started to cry because",
        "correct": " he",
        "incorrect": " she",
    },
]

def logit_gap(model, prompt, correct, incorrect, topk=3):
    input_ids = tokenizer.encode(prompt, return_tensors="pt")
    with torch.no_grad():
        logits = model(input_ids).logits[0, -1]
    correct_id = tokenizer.encode(correct, add_special_tokens=False)[0]
    incorrect_id = tokenizer.encode(incorrect, add_special_tokens=False)[0]
    gap = logits[correct_id].item() - logits[incorrect_id].item()

    top_vals, top_ids = torch.topk(logits, topk)
    top_tokens = [(tokenizer.decode([tid]), val.item()) for tid, val in zip(top_ids, top_vals)]

    return gap, top_tokens

for name in model_names:
    print(f"\n=== {name} ===")
    model = AutoModelForCausalLM.from_pretrained(name)
    model.eval()
    for case in cases:
        gap, top_tokens = logit_gap(model, case["prompt"], case["correct"], case["incorrect"])
        result = "correct" if gap > 0 else "WRONG"
        top_str = ", ".join(f"{repr(t)}={v:.2f}" for t, v in top_tokens)
        print(f"  gap={gap:+.2f}  [{result}]  top-3: {top_str}")
        print(f"    {case['prompt'][:60]}...")
```
